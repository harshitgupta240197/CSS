# CSS
#### CSS (Cascading Style Sheets) is the standard stylesheet language used to design and style the visual layout of web pages. While HTML provides the structural skeleton of a website, CSS acts as the paint, clothing, and interior decoration. It dictates exactly how elements appear on digital screens, paper layouts, or other media interfaces.

#### Docs : https://developer.mozilla.org/en-US/docs/Web/CSS
--------
```css
selector {
  property: value;
}
```
---
```css
h1 {
  color: purple;
}
```
------
------
## CSS Specificity
#### CSS specificity is the set of rules browsers use to determine which CSS property value is applied to an element when multiple competing rules target it. Think of it as a weighting system or scoreboard; the selector with the highest weight wins and applies its styles
--------

## text-decoration
#### The text-decoration property in CSS is a shorthand property used to set the appearance of decorative lines on text. It is most commonly used to add lines (like underlines) or to remove the default underlines from HTML hyperlinks

````python
%%css

h1 {
    text-transform: uppercase;
    font-family: monospace;
    text-align: center;
    font-weight: 100;
    font-size: 40px;
    letter-spacing: 20px;
    text-decoration: underline wavy #e6a8d7 3px;
    text-underline-position: under;
}
````

```css
}

```

# Combining Multiple Selectors
```css

    color : green;
    background-color : black
```

# CLASS Selector
#### Let us assume the name of the class is 'tag'
```css
.tag {
    background-color : black;
    color : white;
}
```
---------

# DESCENDANT SELECTOR
#### Select all ```<a>``` that are nested inside an ```<li>```
#### So in this case the ```<li>``` won't be styled instead the underlying nested anchor tag will be styled
```css
li a {
    color : teal;
    background-color : yellow;
}
```
----

# ADJACENT SELECTOR
#### Select only the paragraphs that are immediately preceded by an ```<h1>```. Paragraphs that come immediately after ```<h1>```
```css
h1 + p {
    color : red
}
```
-----

# DIRECT CHILD
#### Select only the ```<li>``` that are direct children of a ```<div>``` element
```css
div > li {
    color : white;
    background-color : black;
}
```
-------

# ATTRIBUTE SELECTOR
#### Select all input elements where the type attribute is set to 'text
```css
input[type='text'] {
    width : 300px;
    color : yellow;
    background-color : white;
}
```
