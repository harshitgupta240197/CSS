# CSS Box Model
#### The CSS box model is a core structural rule dictating that every visible HTML element is treated as a rectangular box that controls layout, spacing, and sizing

## The Four Layers of a Box
Every element box is comprised of four concentric layers, moving from the inside out:

- **Content**: The core area where text, images, or child elements reside.
- **Padding**: Transparent negative space immediately surrounding the content box. It adopts the background color of the element.
- **Border**: A thin or thick line wrapped directly around the padding.
- **Margin**: The outermost transparent buffer zone that pushes adjacent elements away. It sits outside the border and does not inherit the background color.
