# PSEUDO CLASSES
#### Keyword added to a selector that specifies a specific state of the selected element
#### A CSS pseudo-class is a keyword added to a selector that styles an element based on its temporary state, user interaction, or position in the document tree. Pseudo-classes always begin with a single colon (:)
#### Unlike normal CSS classes that you manually add to HTML (class="active"), pseudo-classes are applied dynamically by the browser
- `:active`
- `:checked`
- `:first`
- `:first-child`
- `:hover`
- `:not()`
- `:nth-child()`
- `:nth-of-type()`
----------
```css
button:hover {
  background-color: darkblue;
}
input:focus {
  border-color: green;
}
```

# Pseudo Elements
#### Keyword added to a selector that lets you style a particular part of selected elemenent(s)
#### A CSS pseudo-element is a keyword added to a selector that allows you to style a specific part of an HTML element. Unlike regular HTML tags, pseudo-elements generate "virtual" structures that do not exist directly in the document object model (DOM)
- `::after`
- `::before`
- `::first-letter`
- `::first-line`
- `::selection`
----
```css
/* Adds a stylized icon before link text */
a::before {
  content: "🔗 ";
  font-size: 1.2rem;
}

/* Creates a custom underline accent via an empty block */
h2::after {
  content: "";
  display: block;
  width: 50px;
  height: 4px;
  background-color: blue;
}
```
